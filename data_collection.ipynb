```python
!pip install -q fiftyone
!rm -rf /root/fiftyone/coco-2017/validation /content/coco3 /content/coco3.zip

import fiftyone as fo, fiftyone.zoo as foz
from fiftyone import ViewField as F

classes = ["person", "dog", "car"]
merged = fo.Dataset("coco3", overwrite=True)

#클래스별 해당 물체가 있는 사진 100장씩 받기
for c in classes:
    ds = foz.load_zoo_dataset(
        "coco-2017", split="validation", label_types=["detections"],
        classes=[c], max_samples=100, shuffle=True, seed=42,
        dataset_name=f"coco_{c}", drop_existing_dataset=True,
    )
    merged.merge_samples(ds)

# 3개 클래스 박스만 남기기
view = merged.filter_labels("ground_truth", F("label").is_in(classes))
print("총 사진 수:", len(view))
print(view.count_values("ground_truth.detections.label"))

# YOLO 형식으로 저장
view.export(export_dir="/content/coco3", dataset_type=fo.types.YOLOv5Dataset,
            label_field="ground_truth", classes=classes, split="train")
!mv /content/coco3/dataset.yaml /content/coco3/data.yaml

# zip으로 묶어
!cd /content && zip -qr coco3.zip coco3
from google.colab import files
files.download("/content/coco3.zip")
```
